# Question 10

```python
alph = input("Enter a single letter:")
check = alph.lower()

if len(alph) == 1:
    if check in ("a","e","i","o","u"):
        print(f"{alph} is a vowel")
    elif check.isalpha():
        print(f"{alph} is a Constant")
    else: 
        print("Enter valid Alphabet")
else:
    print("Enter single letter only please.")
```
