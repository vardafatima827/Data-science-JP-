 A customer purchases 4 shirts at Rs. 1,500 each. Create a function that calculates the total bill.

```python
def cal_bill(quantity,price):
    return quantity * price
```

```python
cal_bill(4,1500)
```

 A cinema only allows customers aged 18 or above. Create a function to check eligibility.

```python
def check_eligibility(age):
    if age >= 18:
        return "You Are Eligible"
    else:
        return "You Are Not Eligible"
```

```python
check_eligibility(19)
```

 An electricity company charges Rs. 15 per unit. Create a function to calculate the electricity bill.

```python
def cal_electicity_bill(units):
    return units * 15
```

```python
cal_electicity_bill(200)
```
