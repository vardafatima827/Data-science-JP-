# Question 1

```python
Balance = 1000
try:
    Amount = int(input("Enter Amount: "))
    if Amount <= 0:
        print("Amount must be more then zero.")
    elif Amount > Balance:
        print("Insufficient Balance.")
    else:
        a = Balance - Amount
        print(f"Remaining Balance is {a} ")
except ValueError:
    print("Please Enter a Valid Number.")
finally:
    print("Thank You for using ATM")
```

# Question 2

```python
available_seats = 5
try:
    Tickets = int(input("How many tickets do you want to book: "))
    if Tickets <= 0:
        print("You must book at least 1 ticket.")
    elif Tickets > available_seats:
        print("Not enough seats available.")
    else:
        print(f"{Tickets} Tickets were booked.")
except ValueError:
    print("Please Enter a Valid Number.")
finally:
    print("Thank you for booking with us.")
```

# Question 7

```python
class Book:
    def __init__(self,title,author,year):
        self.Title = title
        self.Author = author
        self.Year = year
    def info(self):
        print(f"'{self.Title}' by {self.Author}, published in {self.Year}.")
    def is_classic(self):
        print("This is a book.")
```

```python
book = Book("To Kill a Mockingbird","Harper Lee",1960)
```

```python
book.info()
book.is_classic()
```

# Question 8

```python
class students:
    def __init__(self,name,grade, age):
        self.name = name
        self.grade = grade
        self.age = age
    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old.")
    def show_grade(self):
        print(f"My grade is {self.grade}.")
```

```python
stud = students("Alice", "A", 16)
```

```python
stud.introduce()
stud.show_grade()
```

# Question 10

```python
available_slots = [101, 102, 103, 104, 105]
reserved =  [106, 107, 108, 109, 110]
try:
    slot = int(input("Enter the parking slot number you want to reserve: "))
    if slot in available_slots:
        print("Slot {slot} is reserved for you.")
    elif slot in reserved:
        print("Slot {slot} is already reserved by someone else.")
    else:
        print("Slot {slot} does not exist.")
except ValueError:
    print("Invalid input. Please enter a valid slot number.")
```
