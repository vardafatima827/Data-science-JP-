# Question 1

```python
def count_vowels(word):
    c_v = 0
    vowels = ('a', 'e', 'i', 'o', 'u')
    for i in word.lower():
        if i in vowels:
            c_v += 1
    return c_v
    
```

```python
count_vowels("aeioA")
```

# Question 6

```python
def cal_rectangle(lenght,width):
    def cal_area():
        return lenght * width
    def cal_perimeter():
        return 2*(lenght + width)
    area = cal_area()
    perimeter = cal_perimeter()
    print (f"Area : {area}")
    print (f"Perimeter : {perimeter}")
```

```python
cal_rectangle(5,3)
```

# Question 10

```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        total = price * quantity
        return total
    def apply_discount(total):
        return max(0,total - discount) 
    total = calculate_total()
    final_bill = apply_discount(total)
    if discount == 0:
        print(f"Final Bill : {final_bill} no discount")
    else:
        print(f"Final Bill : {final_bill} with discount")  
        
```

```python
shopping_bill(100,6)
```

```python
shopping_bill(100,6,100)
```
