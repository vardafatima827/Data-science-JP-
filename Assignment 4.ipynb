# ***Question 2***

```python
temps = [22, 25, 30, 45, 28] 
critical = 40
for i in temps:
    if i >= critical:
        break
    print(f"Temperature : {i}")
print("Critical temperature reached! Stop monitoring.")
```

# ***Question 6***

```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domain = []
for e in emails:
    d = e.split("@")[1]
    domain.append(d)
print("Domains:", domain)
```
